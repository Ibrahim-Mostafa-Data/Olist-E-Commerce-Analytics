# Python Cleaning Checklist

```python
for i,e in dataframes.items():
    print(f'\n   {i}   ')
# 1. Null Values
    print(f'\n Null:')
    print(e.isnull().sum())
# 2. Duplicate Rows
    print(f'\n Dublicate:')
    print(e.duplicated().sum())
# 3. Data Types
    print(f'\n Types')
    print(e.dtypes)
```

```python
for i,e in dataframes.items():
    Nulls = e.isnull().sum()
    if Nulls.sum() > 0 :
        print(f'\n  {i}  ')
        print(Nulls[Nulls > 0])
```
